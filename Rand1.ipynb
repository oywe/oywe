```python
from PIL import Image
import cv2

img = cv2.imread("./IMG_1880.jpeg")
height, width = img.shape[:2]
img2 = cv2.resize(img, (64, 64))


```

```python
img2
```

```python
img3=cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
img3
```

```python
for i in img3:
  print(i)
```

```python
import hashlib,secrets

hash=hashlib.sha256()

for i in img3:
  for j in i:
    hash.update(int(j).to_bytes(1, 'big'))

hash.update(secrets.randbelow(256).to_bytes(1,"big"))

hash.hexdigest()




```
